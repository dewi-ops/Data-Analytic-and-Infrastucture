```python
def extract_csv():

    df = pd.read_csv(
        "source_uas/siswa_sekolah_a.csv"
    )

    return df
```

```python
def extract_json():

    with open(
        "source_uas/siswa_sekolah_c.json",
        "r",
        encoding="utf-8"
    ) as f:

        data = json.load(f)

    records = []

    for siswa in data["data_siswa"]:

        row = {
            "nisn": siswa["identitas"]["nisn"],
            "nama": siswa["identitas"]["nama"],
            "jenis_kelamin": siswa["identitas"]["jenis_kelamin"],
            "tanggal_lahir": siswa["identitas"]["tanggal_lahir"],

            "sekolah": siswa["sekolah"]["nama"],
            "alamat": siswa["sekolah"]["alamat"],
            "jenjang": siswa["sekolah"]["jenjang"],
            "kota": siswa["sekolah"]["kota"],
            "status": siswa["sekolah"]["status"]
        }

        records.append(row)

    return pd.DataFrame(records)
```

```python
def extract_sql():

    with open(
        "source_uas/siswa_sekolah_b.sql",
        "r",
        encoding="utf-8"
    ) as f:

        sql = f.read()

    pattern = r"\((\d+),\s*'([^']+)',\s*'([^']+)',\s*'([^']+)',\s*'([^']+)',\s*'([^']+)',\s*'([^']+)',\s*'([^']+)',\s*'([^']+)',\s*'([^']+)'\)"

    rows = re.findall(pattern, sql)

    data = []

    for row in rows:

        data.append({
            "nisn": row[1],
            "nama": row[2],
            "jenis_kelamin": row[3],
            "tanggal_lahir": row[4],
            "sekolah": row[5],
            "alamat": row[6],
            "jenjang": row[7],
            "kota": row[8],
            "status": "Aktif"
        })

    return pd.DataFrame(data)
```

```python
df_csv = extract_csv()
df_json = extract_json()
df_sql = extract_sql()
```

```python
print("CSV")
display(df_csv.head())

print("JSON")
display(df_json.head())

print("SQL")
display(df_sql.head())
```

```python
df_csv = df_csv.rename(
    columns={
        "nama_siswa":"nama",
        "alamat_sekolah":"alamat"
    }
)
```

```python
df_csv["status"] = "Aktif"
```

```python
target_columns = [
    'nisn',
    'nama',
    'jenis_kelamin',
    'tanggal_lahir',
    'sekolah',
    'alamat',
    'jenjang',
    'kota',
    'status'
]
```

```python
df_csv = df_csv[target_columns]
df_json = df_json[target_columns]
df_sql = df_sql[target_columns]
```

```python
df_csv["source_file"] = "csv"
df_json["source_file"] = "json"
df_sql["source_file"] = "sql"
```

```python
df_csv["created_at"] = datetime.now()
df_json["created_at"] = datetime.now()
df_sql["created_at"] = datetime.now()
```

```python
master_df = pd.concat(
    [
        df_csv,
        df_json,
        df_sql
    ],
    ignore_index=True
)
```

```python
master_df.drop_duplicates(
    subset=["nisn"],
    inplace=True
)
```

```python
print(master_df.shape)

display(master_df.head(10))
```

```python
import sqlite3

conn = sqlite3.connect(
    "warehouse_siswa.db"
)

cursor = conn.cursor()

print("Database berhasil dibuat")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS master_siswa(

    nisn TEXT PRIMARY KEY,

    nama TEXT,
    jenis_kelamin TEXT,
    tanggal_lahir TEXT,

    sekolah TEXT,
    alamat TEXT,
    jenjang TEXT,
    kota TEXT,
    status TEXT,

    source_file TEXT,
    created_at TEXT

)
""")

conn.commit()

print("Tabel berhasil dibuat")
```

```python
for _, row in master_df.iterrows():
    # Convert created_at Timestamp to ISO formatted string
    row_values = row.tolist()
    row_values[-1] = row_values[-1].isoformat()

    cursor.execute("""
    INSERT INTO master_siswa(

        nisn,
        nama,
        jenis_kelamin,
        tanggal_lahir,

        sekolah,
        alamat,
        jenjang,
        kota,
        status,

        source_file,
        created_at

    )

    VALUES (?,?,?,?,?,?,?,?,?,?,?)

    ON CONFLICT(nisn)
    DO UPDATE SET

        nama=excluded.nama,
        jenis_kelamin=excluded.jenis_kelamin,
        tanggal_lahir=excluded.tanggal_lahir,

        sekolah=excluded.sekolah,
        alamat=excluded.alamat,
        jenjang=excluded.jenjang,
        kota=excluded.kota,
        status=excluded.status,

        source_file=excluded.source_file,
        created_at=excluded.created_at

    """,
    tuple(row_values)
    )

conn.commit()

print("Data berhasil dimasukkan")
```

```python
import schedule
import time

def run_etl_scheduler():

    print("ETL berjalan...")

    logging.info("ETL START")

    try:

        logging.info("ETL SUCCESS")

        print("ETL SUCCESS")

    except Exception as e:

        logging.error(str(e))

        print("ETL FAILED")
```

```python
schedule.every(1).minutes.do(
    run_etl_scheduler
)

print("Scheduler aktif...")
```

```python
while True:

    schedule.run_pending()

    time.sleep(1)
```
